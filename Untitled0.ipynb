```python
import json
from sklearn.model_selection import train_test_split

def normalize(text):
    return str(text).lower().strip()

def get_type(ans):
    ans = ans.lower()
    if ans in ["yes", "no"]:
        return "closed"
    elif len(ans.split()) <= 2:
        return "closed"
    else:
        return "open"
```

```python
def process_vqarad():
    data = json.load(open("data/vqa-rad/train.json"))
    image_dir = "data/vqa-rad/images"

    processed = []

    for item in data:
        try:
            answer = normalize(item["answer"])

            processed.append({
                "image": os.path.join(image_dir, item["image_name"]),
                "question": normalize(item["question"]),
                "answer": answer,
                "type": get_type(answer)
            })
        except:
            continue

    return processed
```

```python
def process_slake():
    val = json.load(open("data/slake/val.json"))
    test = json.load(open("data/slake/test.json"))

    data = val + test
    image_dir = "data/slake/imgs"

    processed = []

    for item in data:
        try:
            # keep only English
            if item.get("q_lang", "en") != "en":
                continue

            answer = normalize(item["answer"])

            # ✅ FIXED LINE
            img_path = os.path.join(image_dir, item["img_name"])

            processed.append({
                "image": img_path,
                "question": normalize(item["question"]),
                "answer": answer,
                "type": get_type(answer)
            })

        except Exception as e:
            continue

    return processed
```

```python
import re

def extract_number(text):
    numbers = re.findall(r"\d+", text)
    if len(numbers) > 0:
        return numbers[0]
    return None
```

```python
def process_pathvqa():
    import json
    import os

    data = json.load(open("data/pathvqa/train.json"))
    image_dir = "data/pathvqa/images"

    processed = []

    for item in data:
        try:
            # clean text
            question = str(item["Questions"]).lower().strip()
            answer = str(item["Answers"]).lower().strip()

            # 🔥 STEP A: extract number
            number = extract_number(item["Image_ID"])

            # skip if no number
            if number is None:
                continue

            # 🔥 STEP B: create correct filename
            img_name = "img_" + number + ".jpg"

            # 🔥 STEP C: full path
            img_path = os.path.join(image_dir, img_name)

            # 🔥 STEP D: check file exists
            if not os.path.exists(img_path):
                continue

            # save
            processed.append({
                "image": img_path,
                "question": question,
                "answer": answer,
                "type": "closed" if answer in ["yes", "no"] else "open"
            })

        except:
            continue

    return processed
```

```python
import torch

def load_glove(path):
    glove = {}
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            vector = torch.tensor([float(x) for x in values[1:]])
            glove[word] = vector
    return glove

glove_path = "/content/drive/MyDrive/MedicalVQA/glove.6B.300d.txt"
glove = load_glove(glove_path)
```

```python
import re

def preprocess_question(q):
    q = q.lower()
    q = re.sub(r"[^a-z0-9 ]", "", q)
    return q.split()

def glove_question_embedding(question):
    tokens = preprocess_question(question)

    vectors = []
    for t in tokens:
        if t in glove:
            vectors.append(torch.tensor(glove[t]))

    if len(vectors) == 0:
        return torch.zeros(300)

    return torch.stack(vectors).mean(dim=0)
```

```python
from torch.utils.data import Dataset
from PIL import Image

class MedicalVQADataset(Dataset):
    def __init__(self, json_path):
        self.data = json.load(open(json_path))

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        item = self.data[idx]

        image = Image.open(item["image"]).convert("RGB")
        image = transform(image)

        question_vec = glove_question_embedding(item["question"])

        return {
            "image": image,
            "question_vec": question_vec,
            "question": item["question"],
            "answer": item["answer"],
            "type": item["type"]
        }
```

```python
def collate_fn(batch):
    return {
        "image": torch.stack([x["image"] for x in batch]),
        "question_vec": torch.stack([x["question_vec"] for x in batch]),
        "question": [x["question"] for x in batch],
        "answer": [x["answer"] for x in batch],
        "type": [x["type"] for x in batch]
    }
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import re

import timm

from torch.utils.data import DataLoader
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv, global_mean_pool
from sklearn.neighbors import NearestNeighbors
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

```python
def load_glove(path):
    glove = {}
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            vector = np.array(values[1:], dtype=np.float32)
            glove[word] = vector
    return glove

glove_path = "/content/drive/MyDrive/MedicalVQA/glove.6B.300d.txt"
glove = load_glove(glove_path)

print("Loaded words:", len(glove))
```

```python
def preprocess_question(q):
    q = q.lower()
    q = re.sub(r"[^a-z0-9 ]", "", q)
    return q.split()

def question_to_vec(question, dim=300):
    tokens = preprocess_question(question)

    vecs = []
    for t in tokens:
        if t in glove:
            vectors.append(torch.tensor(glove[t]))

    if len(vecs) == 0:
        return torch.zeros(dim)

    return torch.tensor(vecs).mean(dim=0)
```

```python
swin = timm.create_model(
    'swin_tiny_patch4_window7_224',
    pretrained=True
)

swin.head = nn.Identity()
swin = swin.to(device)
swin.eval()
```

```python
def extract_patches(images):
    with torch.no_grad():
        feats = swin.forward_features(images)
    return feats  # [B, 49, 768]
```

```python
def build_graph(node_feats, k=5):
    feats = node_feats.cpu().numpy()

    nbrs = NearestNeighbors(n_neighbors=k+1).fit(feats)
    _, idxs = nbrs.kneighbors(feats)

    edges = []
    for i in range(len(feats)):
        for j in idxs[i]:
            if i != j:
                edges.append([i, j])
                edges.append([j, i])

    return torch.tensor(edges).t().contiguous()
```

```python
class QGNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.node_proj = nn.Linear(768, 300)
        self.q_proj = nn.Linear(300, 300)

        self.gcn1 = GCNConv(600, 256)
        self.gcn2 = GCNConv(256, 256)

    def forward(self, x, edge_index, q_embed):
        x = self.node_proj(x)

        q = self.q_proj(q_embed)
        q_expand = q.unsqueeze(0).expand(x.size(0), -1)

        x = torch.cat([x, q_expand], dim=1)

        x = F.relu(self.gcn1(x, edge_index))
        x = self.gcn2(x, edge_index)

        return x
```

```python
class ClosedModel(nn.Module):
    def __init__(self, qgnn):
        super().__init__()
        self.qgnn = qgnn
        self.fc = nn.Linear(256, 2)

    def forward(self, data):
        node_emb = self.qgnn(data.x, data.edge_index, data.q_embed)

        batch = torch.zeros(node_emb.size(0), dtype=torch.long).to(device)
        graph_emb = global_mean_pool(node_emb, batch)

        return self.fc(graph_emb)
```

```python
def extract_keywords(node_emb, top_k=3):
    importance = node_emb.mean(dim=1)
    top_idx = torch.topk(importance, top_k).indices

    return [f"region_{i.item()}" for i in top_idx]
```

```python
def generate_answer(keywords):
    return f"Image shows {keywords[0]} and {keywords[1]}"
```

```python
from tqdm import tqdm

optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
criterion = nn.CrossEntropyLoss()

epochs = 10

for epoch in range(epochs):
    model.train()

    total_loss = 0
    correct = 0
    total = 0

    pbar = tqdm(vqa_loader, desc=f"Epoch {epoch+1}")

    for batch in pbar:

        images = batch["image"].to(device)
        q_vec = batch["question_vec"].to(device)
        types = batch["type"]
        answers = batch["answer"]

        patches = extract_patches(images)

        for i in range(len(images)):

            if types[i] != "closed":
                continue

            x = patches[i].reshape(-1, patches.shape[-1])
            q = q_vec[i]

            edge_index = build_graph(x)

            data = Data(x=x, edge_index=edge_index)
            data.q_embed = q

            label = 1 if answers[i] == "yes" else 0
            label = torch.tensor([label]).to(device)

            optimizer.zero_grad()

            logits = model(data)
            loss = criterion(logits, label)

            loss.backward()
            optimizer.step()

            total_loss += loss.item()

            pred = torch.argmax(logits, dim=1)
            correct += (pred == label).sum().item()
            total += 1

        # 🔥 LIVE UPDATE
        if total > 0:
            pbar.set_postfix({
                "loss": f"{total_loss:.2f}",
                "acc": f"{correct/total:.3f}"
            })

    epoch_acc = correct / total if total > 0 else 0

    print(f"\nEpoch {epoch+1} Summary")
    print("Total Loss:", total_loss)
    print("Accuracy:", epoch_acc)
```

```python
# Map node positions to pseudo-anatomical regions
def map_to_region(idx):
    regions = [
        "upper region", "lower region", "central region",
        "left side", "right side",
        "anterior area", "posterior area"
    ]
    return regions[idx % len(regions)]


def extract_keywords(node_emb, top_k=3):
    scores = node_emb.norm(dim=1)
    top_idx = torch.topk(scores, top_k).indices

    keywords = [map_to_region(i.item()) for i in top_idx]
    return keywords
```

```python
def generate_answer(keywords, question):
    q = question.lower()

    if "what" in q:
        return f"The image indicates abnormality in the {keywords[0]}."

    elif "where" in q:
        return f"The finding is located in the {keywords[0]}."

    elif "how many" in q:
        return f"Multiple regions are involved, including the {keywords[0]} and {keywords[1]}."

    else:
        return f"Relevant areas include the {', '.join(keywords)}."
```
